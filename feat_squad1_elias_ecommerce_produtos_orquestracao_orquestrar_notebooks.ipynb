# Master Orchestrator — Camadas Bronze → Silver → Gold (Produtos)

Este notebook orquestra a execução sequencial dos três notebooks da pipeline de produtos:

1. **Bronze** — `feat_squad1_elias_ecommerce_produtos_bronze_implementar_bronze`
2. **Silver**  — `feat_squad1_elias_ecommerce_produtos_silver_implementar_silver`
3. **Gold**   — `feat_squad1_elias_ecommerce_produtos_gold_implementar_gold`

Cada notebook é executado em sequência. Se algum falhar, a execução é interrompida imediatamente.

> **Pré‑requisito**: Todos os notebooks devem estar no mesmo diretório (ou no repositório `merca-data-platform-produtos`) para que o caminho relativo funcione.


```python
# ─── Configurações ─────────────────────────────────────────────────────────────

# Nomes dos notebooks (sem extensão .py ou .ipynb – o Databricks infere)
NOTEBOOK_BRONZE = "feat_squad1_elias_ecommerce_produtos_bronze_implementar_bronze"
NOTEBOOK_SILVER = "feat_squad1_elias_ecommerce_produtos_silver_implementar_silver"
NOTEBOOK_GOLD   = "feat_squad1_elias_ecommerce_produtos_gold_implementar_gold"

# Timeout máximo para cada notebook (em segundos)
TIMEOUT = 3600   # 1 hora por etapa

# Parâmetros opcionais (nenhum por enquanto)
PARAMS = {}

print("🚀 Iniciando orquestração da pipeline de PRODUTOS")
print(f"   Bronze : {NOTEBOOK_BRONZE}")
print(f"   Silver : {NOTEBOOK_SILVER}")
print(f"   Gold   : {NOTEBOOK_GOLD}")
```

```python
# ─── Função auxiliar para executar um notebook e validar o resultado ─────────

def executar_notebook(nome_notebook, timeout, params):
    """
    Executa um notebook via dbutils.notebook.run e verifica se teve sucesso.
    Lança uma exceção se falhar.
    """
    print(f"\n{'='*70}")
    print(f"▶ Executando: {nome_notebook}")
    print(f"{'='*70}")
    
    try:
        resultado = dbutils.notebook.run(nome_notebook, timeout, params)
        # O retorno é uma string com a mensagem de saída do notebook.
        # Se o notebook chamar dbutils.notebook.exit("mensagem"), isso é retornado.
        # Se houver erro, uma exceção será lançada.
        print(f"✅ {nome_notebook} concluído com sucesso.")
        print(f"   Saída: {resultado}")
        return resultado
    except Exception as e:
        print(f"❌ Falha ao executar {nome_notebook}: {e}")
        raise  # Propaga a exceção para interromper a execução
```

```python
# ─── 1. Bronze ──────────────────────────────────────────────────────────────────

executar_notebook(NOTEBOOK_BRONZE, TIMEOUT, PARAMS)
```

```python
# ─── 2. Silver ──────────────────────────────────────────────────────────────────

executar_notebook(NOTEBOOK_SILVER, TIMEOUT, PARAMS)
```

```python
# ─── 3. Gold ────────────────────────────────────────────────────────────────────

executar_notebook(NOTEBOOK_GOLD, TIMEOUT, PARAMS)
```

```python
# ─── Finalização ──────────────────────────────────────────────────────────────

print("\n" + "="*70)
print("🎉 PIPELINE DE PRODUTOS CONCLUÍDA COM SUCESSO!")
print("   Bronze → Silver → Gold executados em sequência.")
print("="*70)
```
