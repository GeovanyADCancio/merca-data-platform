```python
#verificar se a tabela da lista existe ou não

tabelas = [
    "bronze_ecommerce_clientes",
    "bronze_ecommerce_enderecos",
    "bronze_ecommerce_itens_pedido",
    "bronze_ecommerce_pedidos",
    "bronze_ecommerce_rastreamento",
    "silver_ecommerce_clientes",
    "silver_ecommerce_enderecos",
    "silver_ecommerce_itens_pedido",
    "silver_ecommerce_pedidos",
    "silver_ecommerce_rastreamento",
    "dq_monitoring_logs",
]

for tabela in tabelas:
    nome = f"squad1.{tabela}"
    existe = spark.catalog.tableExists(nome)
    print(f"{nome}: {'EXISTE' if existe else 'NÃO EXISTE'}")
```
